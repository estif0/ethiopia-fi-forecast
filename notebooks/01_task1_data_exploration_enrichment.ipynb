## 2. Data Loading and Schema Understanding

Load the unified dataset and reference codes using the DataLoader class.

### Understanding Record Types

The unified schema uses `record_type` to differentiate between different kinds of data:

```python
# View sample of each record type
for record_type in ['observation', 'event', 'target']:
    print(f"\n{'='*70}")
    print(f"Sample {record_type.upper()} records:")
    print('='*70)
    sample = loader.get_records_by_type(record_type)
    if len(sample) > 0:
        display(sample.head(3))
    else:
        print(f"No {record_type} records found")
```

## 3. Exploratory Analysis

### 3.1 Indicators Coverage

```python
# List all unique indicators
indicators = loader.get_indicators()
observations = loader.get_records_by_type('observation')

print(f"Total Unique Indicators: {len(indicators)}\n")
print("Indicator Coverage:")
for ind in indicators[:10]:  # Show first 10
    count = len(observations[observations['indicator_code'] == ind])
    print(f"  - {ind}: {count} observations")
    
if len(indicators) > 10:
    print(f"  ... and {len(indicators) - 10} more indicators")
```

### 3.2 Events Analysis

```python
# Examine events
events = loader.get_records_by_type('event')

print(f"Total Events: {len(events)}\n")

if 'category' in events.columns:
    print("Events by Category:")
    print(events['category'].value_counts())
    
print(f"\n{'='*70}")
print("Sample Events:")
print('='*70)
if len(events) > 0:
    display(events[['record_id', 'category', 'indicator', 'observation_date']].head())
```

## 4. Enriched Dataset Verification

Load and verify the enriched dataset that was created through the enrichment process.

```python
# Load enriched dataset
enriched_data = pd.read_csv('../data/processed/ethiopia_fi_enriched.csv')

print(f"Original dataset: {len(unified_data)} records")
print(f"Enriched dataset: {len(enriched_data)} records")
print(f"New records added: {len(enriched_data) - len(unified_data)}")

print("\nEnriched Dataset Record Type Distribution:")
print(enriched_data['record_type'].value_counts())
```

```python
# Check new observations added
new_observations = enriched_data[
    ~enriched_data['record_id'].isin(unified_data['record_id']) & 
    (enriched_data['record_type'] == 'observation')
]

print(f"\nNew Observations Added: {len(new_observations)}")
if len(new_observations) > 0:
    print("\nNew Indicators Added:")
    display(new_observations[['record_id', 'indicator', 'indicator_code', 'value_numeric']].head(10))
```
